```python
import pandas as pd
import joblib
```

```python
#load the best model and  test models
best_model = joblib.load('../Models/best_model.pkl')
x_test = joblib.load('../Models/x_test.pkl')
```

```python
def predict_cancer(input_data):
    input_df = pd.DataFrame(
        input_data,
        columns=x_test.columns
    )
    prediction = best_model.predict(input_df)
    if prediction[0] == 0:
        print("Diagnosis: Tumor is Benign")
    else:
        print("Diagnosis: Tumor is Malignant")
```

```python
input_data = x_test.iloc[0].values.reshape(1, -1)
print("sample input:",input_data)
```

```python
predict_cancer(input_data)
```
